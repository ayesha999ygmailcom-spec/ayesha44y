## 1. Load the McDonald's Nutrition Dataset
We will create a comprehensive representative dataset containing nutrition facts for various McDonald's menu items across different categories.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Creating a representative McDonald's Nutrition Dataset
data = {
    'Category': ['Breakfast', 'Breakfast', 'Breakfast', 'Beef & Pork', 'Beef & Pork', 'Chicken & Fish', 'Chicken & Fish', 'Salads', 'Salads', 'Snacks & Sides', 'Snacks & Sides', 'Desserts', 'Desserts', 'Beverages', 'Beverages'],
    'Item': ['Egg McMuffin', 'Sausage McMuffin', 'Hash Brown', 'Big Mac', 'Quarter Pounder with Cheese', 'McChicken', 'Filet-O-Fish', 'Premium Bacon Ranch Salad', 'Premium Southwest Salad', 'Small French Fries', 'Medium French Fries', 'Baked Apple Pie', 'Vanilla Cone', 'Coca-Cola (Small)', 'McCafe Latte (Small)'],
    'Calories': [300, 370, 150, 530, 520, 360, 390, 230, 350, 230, 340, 250, 170, 140, 120],
    'Total_Fat_g': [13.0, 23.0, 9.0, 27.0, 26.0, 15.0, 19.0, 7.0, 11.0, 11.0, 16.0, 13.0, 3.5, 0.0, 4.5],
    'Sodium_mg': [750, 780, 310, 960, 1100, 800, 590, 430, 850, 130, 190, 170, 70, 0, 115],
    'Protein_g': [17, 14, 2, 24, 30, 14, 15, 9, 12, 2, 4, 2, 5, 0, 10]
}

df = pd.DataFrame(data)
print("Dataset loaded successfully. Shape:", df.shape)
df.head(15)
```

## 2. Descriptive Summary Statistics
Let's evaluate the central tendencies, spread, and distributions of the numeric variables.

```python
print("### Dataset Info ###")
df.info()
print("\n### Statistical Summary ###")
df.describe()
```

## 3. Exploratory Data Visualization
We will visualize distributions, group comparisons, and feature correlations.

```python
plt.figure(figsize=(15, 10))

# 1. Calories Distribution
plt.subplot(2, 2, 1)
sns.histplot(df['Calories'], kde=True, color='skyblue')
plt.title('Distribution of Calories')

# 2. Calories by Category
plt.subplot(2, 2, 2)
sns.boxplot(data=df, x='Calories', y='Category', palette='Set2')
plt.title('Calories by Menu Category')

# 3. Scatter Plot: Calories vs. Protein
plt.subplot(2, 2, 3)
sns.scatterplot(data=df, x='Calories', y='Protein_g', hue='Category', s=100)
plt.title('Protein vs. Calories')

# 4. Correlation Heatmap
plt.subplot(2, 2, 4)
numeric_cols = df.select_dtypes(include=[np.number])
sns.heatmap(numeric_cols.corr(), annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Nutritional Correlation Matrix')

plt.tight_layout()
plt.show()
```

## 4. Model Preparation: Train-Test Split and Decision Tree
We will now prepare our features (`Total_Fat_g`, `Sodium_mg`, `Protein_g`) to predict `Calories` using a Decision Tree Regressor, splitting the data into 80% training and 20% testing sets.

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import mean_squared_error, r2_score
from sklearn import tree

# Define features (X) and target (y)
X = df[['Total_Fat_g', 'Sodium_mg', 'Protein_g']]
y = df['Calories']

# Split the dataset into 80% training and 20% testing
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)

print(f"Training features shape: {X_train.shape}")
print(f"Testing features shape: {X_test.shape}")
```

### Train the Decision Tree Regressor and Evaluate Performance
Let's fit the model and evaluate the Mean Squared Error (MSE) and R-squared ($R^2$) score on the test set.

```python
# Initialize and train the Decision Tree Regressor
dt_model = DecisionTreeRegressor(random_state=42, max_depth=3)
dt_model.fit(X_train, y_train)

# Predict on the test set
y_pred = dt_model.predict(X_test)

# Evaluation Metrics
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Mean Squared Error (MSE): {mse:.2f}")
print(f"R-squared (R2) Score: {r2:.2f}")
```

### Visualize the Decision Tree Structure

```python
plt.figure(figsize=(12, 8))
tree.plot_tree(dt_model, feature_names=X.columns, filled=True, rounded=True)
plt.title("Trained Decision Tree Regressor (Predicting Calories)")
plt.show()
```

## 5. Predicting Calorie Content for a Custom Menu Item
We can use our trained AI model to predict the calories of a custom or new menu item based on its macronutrient breakdown.

```python
# Define a custom menu item: Fat = 15g, Sodium = 500mg, Protein = 12g
custom_item = pd.DataFrame([{
    'Total_Fat_g': 15.0,
    'Sodium_mg': 500.0,
    'Protein_g': 12.0
}])

predicted_calories = dt_model.predict(custom_item)
print(f"Predicted Calories for custom item: {predicted_calories[0]:.2f} kcal")
```

## 6. Training an Alternative Model: Random Forest Regressor
To try and improve the prediction performance, we can train a Random Forest Regressor.

```python
from sklearn.ensemble import RandomForestRegressor

# Initialize and train Random Forest model
rf_model = RandomForestRegressor(n_estimators=50, random_state=42)
rf_model.fit(X_train, y_train)

# Predict and evaluate
y_pred_rf = rf_model.predict(X_test)
mse_rf = mean_squared_error(y_test, y_pred_rf)
r2_rf = r2_score(y_test, y_pred_rf)

print(f"Random Forest - Mean Squared Error (MSE): {mse_rf:.2f}")
print(f"Random Forest - R-squared (R2) Score: {r2_rf:.2f}")
```

## 7. Model Results and Accuracy Comparison

Here is a summary of the prediction results and accuracy metrics for the trained AI models predicting calorie values:

| Model | Mean Squared Error (MSE) | R-squared ($R^2$) Score |
| :--- | :---: | :---: |
| **Decision Tree Regressor** | 8966.67 | -9.35 |
| **Random Forest Regressor** | 3432.40 | -2.96 |

### Interpretation of the Results:
1. **Mean Squared Error (MSE)**: Measures the average squared difference between the estimated values and the actual value. Lower is better. The **Random Forest model performed significantly better**, cutting the MSE down by over half from **8966.67** to **3432.40**.
2. **R-squared ($R^2$)**: Represents the proportion of variance in the dependent variable that is predictable from the independent variables. A negative $R^2$ indicates that a simple baseline mean predictor would perform better than the model on this specific split. This occurs because of our extremely small test size (only 3 items).

To evaluate our model performance at an individual item level, let's select beverages and a mix of sample items, feed them to both the Decision Tree and Random Forest models, and display the actual vs. predicted values.

We will save our trained `dt_model` (Decision Tree) and `rf_model` (Random Forest) models as `.pkl` files using the `pickle` library.
